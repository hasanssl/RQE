```python
import pickle
import gzip

split="test"
fl="90"
with gzip.open(f"./data/gzip_signbd/signbd_{fl}.{split}", 'rb') as f:
    # print(f"./data/gzip_signbd/signbd_{fl}.{split}")
    annotations_org = pickle.load(f)
    annotations_sf=[]
f.close()
no_items=len(annotations_org)
for count in range(no_items):
    # print(annotations_org[count])
    item =annotations_org[count]
    sample = {}
    sample['name'] = item['name']
    sample['signer'] = item['signer']
    sample['gloss'] = item['gloss'] 
    sample['text'] = item['text']
    sf_224=item['sign']
    sf_224.reshape(-1,224)
    sf_224[:,33:39]=0
    sample['sign'] = sf_224
    annotations_sf.append(sample)
        
        


with gzip.open(f"./data/gzip_signbd/ShoulderFreeze/signbd_{fl}.{split}", "wb") as file:
    pickle.dump(annotations_sf, file)
file.close()
```
