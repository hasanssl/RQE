```python
import pickle
import gzip


with gzip.open("./data/WLASL_2000_RAW/WLASL_2000.train", 'rb') as f:
    annotations_org = pickle.load(f)
f.close()
no_items=len(annotations_org)
for count in range(no_items):
    # print(annotations_org[count])
    item =annotations_org[count]
    sample = {}
    sample['name'] = item['name']+"_2"
    sample['signer'] = item['signer']
    sample['gloss'] = item['gloss'] 
    sample['text'] = item['text']
    sample['sign'] = item['sign']
    annotations_org.append(sample)
    # print(sample)
        
        


with gzip.open(r"/media/cse/TOSHIBA EXT/WLASL_2000_DUP.train", "wb") as file:
    pickle.dump(annotations_org, file)
file.close()
```
