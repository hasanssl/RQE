```python
import numpy as np
import torch
import tensorflow as tf
```

```python
tf.test.is_gpu_available()
```

```python
PER_FRAME_FEATURE = 1629
def extractPoseHandFace(npy):
    np_matrix=npy.reshape(-1,PER_FRAME_FEATURE)

    pose  = np_matrix[: ,0:94]
    facehands  = np_matrix[: ,99:PER_FRAME_FEATURE]
    
 
    arr = np.concatenate((pose, facehands), axis=1)


    return arr
```

```python
def getNPY_1624(split, name):
   file_name =  f"../relativeQ/AUTSL226/{split}/{name}" 
    
   npy = np.load(file_name)
   npy = extractPoseHandFace(npy)
   return torch.from_numpy(npy)
```

```python
import pickle
import gzip
import os

# dict_keys(['name', 'signer', 'gloss', 'text', 'sign'])

# Test
test_path  = "../relativeQ/AUTSL226/test"
annotations = []   

for trial in os.listdir(test_path):
    sample = {}

    sample['name'] = "test/"+trial
    sample['signer'] = trial.split("_")[1]
    sample['gloss'] = trial.split("_")[0]
    sample['text'] = trial.split("_")[0]
    sample['sign'] = getNPY_1624("test", trial)

    annotations.append(sample)
    
# save the gzipped file
with gzip.open("../gzip_dataset/AUTSL226/AUTSL226_1624.test", "wb") as file:
    pickle.dump(annotations, file)
file.close()
```
