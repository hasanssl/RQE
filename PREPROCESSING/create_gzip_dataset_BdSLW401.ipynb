```python
PER_FRAME_FEATURE = 1629
def extractPoseHand(npy):
    np_matrix=npy.reshape(-1,PER_FRAME_FEATURE)

    pose  = np_matrix[: ,0:98]
    hands  = np_matrix[: ,1503:PER_FRAME_FEATURE]
    
 
    arr = np.concatenate((pose, hands), axis=1)


    return arr
```

```python
import os
def getNPY_224(base, split, name):
   file_name=os.path.join(base, split,name)
    
   npy = np.load(file_name)
   npy = extractPoseHand(npy)
   return torch.from_numpy(npy)
```

```python
import pickle
import gzip
import os

# dict_keys(['name', 'signer', 'gloss', 'text', 'sign'])

# Test
tensorPath  = r"H:\relativeQ\BdSLW401\Front"
gzipPath=r"H:\gzip_BdSLW401\MissingWristHandled"

# tensorPath  = r"H:\NumericalData_BdSLW401\Lateral"
# gzipPath=r"H:\gzip_BdSLW401\RAW"
 
mlsplits=os.listdir(tensorPath)
frontLateral=tensorPath.split('\\')[-1]
for mlsplit in mlsplits:
    annotations = []  
    for trial in os.listdir(os.path.join(tensorPath,mlsplit)):
        sample = {}

        sample['name'] = f"{frontLateral}/{mlsplit}/"+trial
        sample['signer'] = trial[4:7]
        sample['gloss'] = trial[0:4]
        sample['text'] = trial[0:4]
        
        sample['sign'] = getNPY_224(tensorPath, mlsplit, trial)
        # print(sample)
        annotations.append(sample)
    
    # save the gzipped file
    gpath=os.path.join(gzipPath,f"BdSLW401_{frontLateral}.{mlsplit}")
    with gzip.open(gpath, "wb") as file:
        pickle.dump(annotations, file)
        file.close()






```
