```python
PER_FRAME_FEATURE = 1629
def extractPoseHand(npy):
    np_matrix=npy.reshape(-1,PER_FRAME_FEATURE)

    pose  = np_matrix[: ,0:98]
    hands  = np_matrix[: ,1503:PER_FRAME_FEATURE]
    
 
    arr = np.concatenate((pose, hands), axis=1)


    return arr
```

```python
import os
def getNPY_224(base, split, name):
   file_name=os.path.join(base, split,name)
    
   npy = np.load(file_name)
   npy = extractPoseHand(npy)
   return torch.from_numpy(npy)
```

```python
import pickle
import gzip
import os

# dict_keys(['name', 'signer', 'gloss', 'text', 'sign'])

# Test
# tensorPath  = r"H:\relativeQ\90"
tensorPath  = r"H:\NumericalData_signbd\200"
gzipPath=r"H:\gzip_signbd\RAW"


 
mlsplit="test"

annotations = []  
for trial in os.listdir(os.path.join(tensorPath,mlsplit)):
    sample = {}
    # p5_c_aam
    sp=(trial.split('.')[0]).split('_')
    sample['name'] = trial
    sample['signer'] = sp[0]
    sample['gloss'] = sp[-1]
    sample['text'] = sp[-1]
        
    sample['sign'] = getNPY_224(tensorPath, mlsplit, trial)
    
    annotations.append(sample)
    
    # save the gzipped file
gpath=os.path.join(gzipPath,f"signbd_.{mlsplit}")
with gzip.open(gpath, "wb") as file:
    pickle.dump(annotations, file)
    file.close()






```

```python
import random
mlsplit="train"
dict={}
trainlist=os.listdir(os.path.join(tensorPath,mlsplit))
for trial in trainlist:
    sp=(trial.split('.')[0]).split('_')[-1]
    try:
        dict[sp]=dict[sp]+1
    except:

        dict[sp]=1
print(dict)
for key in dict:
    dict[key] = int(dict[key]*0.125 +0.5)
print(dict)

random.shuffle(trainlist)

annotations = []  
val_tensor=[]
for trial in trainlist:
    sample = {}
    # p5_c_aam
    sp=(trial.split('.')[0]).split('_')
    gloss=sp[-1]
    sample['name'] = trial
    sample['signer'] = sp[0]
    sample['gloss'] = gloss
    sample['text'] = gloss
        
    sample['sign'] = getNPY_224(tensorPath, mlsplit, trial)

    if dict[gloss] >=1:
        dict[gloss] = dict[gloss]-1
        val_tensor.append(sample)
    else:
        annotations.append(sample)
    
    # save the gzipped file
gpath=os.path.join(gzipPath,f"signbd_.{mlsplit}")
with gzip.open(gpath, "wb") as file:
    pickle.dump(annotations, file)
    file.close()

gpath=os.path.join(gzipPath,"signbd_.val")
with gzip.open(gpath, "wb") as file:
    pickle.dump(val_tensor, file)
    file.close()
```
