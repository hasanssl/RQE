```python
import numpy as np
import torch
import pickle
import gzip
```

```python
PER_FRAME_FEATURE = 1629
```

```python
def extractPoseHand(npy):
    np_matrix=npy.reshape(-1,PER_FRAME_FEATURE)

    pose  = np_matrix[: ,0:98]
    hands  = np_matrix[: ,1503:PER_FRAME_FEATURE]
    
 
    arr = np.concatenate((pose, hands), axis=1)


    return arr
```

```python
def getNPY_224(name):
   file_name =  "../relativeQ/PHOENIX-2014-T/" + name + ".npy"
   
    
   npy = np.load(file_name)
   npy = extractPoseHand(npy)
   return torch.from_numpy(npy)
```

```python
data_path = "../data/PHOENIX2014T/"
save_path = "../gzip_dataset/PHOENIX2014T/"
```

```python
import pickle
import gzip


with gzip.open(data_path + "phoenix14t.pami0.dev", 'rb') as f:
    annotations_dev = pickle.load(f)
   
    for trial in annotations_dev:
        # print(trial.keys())
        # print(trial['sign'])
        
        npy_tensor = getNPY_224(trial['name'])
        # print(npy_tensor)

        trial['sign'] = npy_tensor

f.close()
with gzip.open(save_path + "phoenix14t.pami0.dev", "wb") as file:
    pickle.dump(annotations_dev, file)
file.close()
```

```python
with gzip.open(data_path + "phoenix14t.pami0.test", 'rb') as f:
    annotations_test = pickle.load(f)
   
    for trial in annotations_test:
        # print(trial.keys())
        # print(trial['sign'])
        
        npy_tensor = getNPY_224(trial['name'])
        # print(npy_tensor)

        trial['sign'] = npy_tensor

f.close()

with gzip.open(save_path+ "phoenix14t.pami0.test", "wb") as file:
    pickle.dump(annotations_test, file)
file.close()
```

```python
with gzip.open(data_path + "phoenix14t.pami0.train", 'rb') as f:
    annotations_train = pickle.load(f)
   
    for trial in annotations_train:
        # print(trial.keys())
        # print(trial['sign'])
        
        npy_tensor = getNPY_224(trial['name'])
        # print(npy_tensor)

        trial['sign'] = npy_tensor

f.close()

with gzip.open(save_path + "phoenix14t.pami0.train", "wb") as file:
    pickle.dump(annotations_train, file)
file.close()
```
